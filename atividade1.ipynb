```python
from sklearn.linear_model import LinearRegression
duracoes = [[10], [15], [20], [25],[30], [35], [40], [50], [55], [60]]
precos = [10,13,14,18,19,23,24,30,32,34] 
modelo = LinearRegression()
modelo.fit(duracoes,precos)
previsao_45 = modelo.predict([[45]])
print("preço estimado para 45 minutos: s$", round(previsao_45[0], 2 ))

previsao_20 = modelo.predict ([[20]])
previsao_60 = modelo.predict([[60]])

print("preço estimado para 20 minutos: r$", round(previsao_20[0], 2))
print("preço estimado para 60 minutos: r$", round(previsao_60[0], 2))
```
